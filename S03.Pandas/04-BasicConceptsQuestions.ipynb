```python
# A Series can be created in multiple ways:

# Method 1: From a list
series1 = pd.Series([1, 2, 3], index=['a', 'b', 'c'])
print(series1)

# Method 2: From a dictionary
data = {'a': 1, 'b': 2, 'c': 3}
series2 = pd.Series(data)
print(series2)

# Method 3: From a NumPy array
array = np.array([10, 20, 30])
series3 = pd.Series(array, index=['x', 'y', 'z'])
print(series3)

# Method 4: From a scalar with an index
series4 = pd.Series(5, index=['a', 'b', 'c'])
print(series4)

# Output (for all methods):
# a    1
# b    2
# c    3
# dtype: int64 (or similar for other examples)
```

```python
# Missing data (NaN, None) can be handled using various methods:

# Method 1: Detect missing values
df = pd.DataFrame({'A': [1, np.nan, 3], 'B': [4, 5, np.nan]})
print(df,"\n")
print(df.isnull())  # Check for missing values
# Output:
#        A      B
# 0  False  False
# 1   True  False
# 2  False   True

# Method 2: Drop missing values
print(df.dropna())  # Drop rows with any missing values
# Output:
#      A    B
# 0  1.0  4.0

# Method 3: Fill missing values
print(df.fillna(0))  # Fill NaN with 0
# Output:
#      A    B
# 0  1.0  4.0
# 1  0.0  5.0
# 2  3.0  0.0

# Method 4: Fill with mean
print(df.fillna(df.mean()))  # Fill NaN with column mean
# Output:
#      A    B
# 0  1.0  4.0
# 1  2.0  5.0
# 2  3.0  4.5

# Method 5: Interpolate missing values
print(df.interpolate())  # Linear interpolation
# Output:
#      A    B
# 0  1.0  4.0
# 1  2.0  5.0
# 2  3.0  5.0
```

#### **10. What is indexing in Pandas?**

```python
# Indexing in Pandas refers to selecting subsets of data using labels or positions.
# - Series: Uses a single index (labels or integers).
# - DataFrame: Uses row and column indices (labels or integers).

# Example:
df = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]}, index=['x', 'y', 'z'])

# Label-based indexing with loc
print(df.loc['x'])  # Select row 'x'
# Output:
# A    1
# B    4
# Name: x, dtype: int64

# Position-based indexing with iloc
print(df.iloc[0])  # Select first row
# Output:
# A    1
# B    4
# Name: x, dtype: int64

# Setting a custom index
df.set_index('A', inplace=True)
print(df)
# Output:
#    B
# A   
# 1  4
# 2  5
# 3  6
```

```python
# Pandas provides methods to detect missing values (NaN, None):

# Method 1: Use isnull()
df = pd.DataFrame({'A': [1, np.nan, 3], 'B': [4, 5, np.nan]})
print(df.isnull())
# Output:
#        A      B
# 0  False  False
# 1   True  False
# 2  False   True

# Method 2: Use notnull()
print(df.notnull())
# Output:
#        A      B
# 0   True   True
# 1  False   True
# 2   True  False

# Method 3: Count missing values per column
print(df.isnull().sum())
# Output:
# A    1
# B    1
# dtype: int64

# Method 4: Total missing values
print(df.isnull().sum().sum())
# Output: 2
```

```python
# Use dropna() to remove rows (or columns) with missing values:

# Method 1: Drop rows with any missing values
df = pd.DataFrame({'A': [1, np.nan, 3], 'B': [4, 5, np.nan]})
print(df.dropna())
# Output:
#      A    B
# 0  1.0  4.0

# Method 2: Drop rows with all missing values
df = pd.DataFrame({'A': [np.nan, np.nan, 3], 'B': [np.nan, 5, np.nan]})
print(df.dropna(how='all'))
# Output:
#      A    B
# 1  NaN  5.0
# 2  3.0  NaN

# Method 3: Drop rows with missing values in specific columns
print(df.dropna(subset=['B']))
# Output:
#      A    B
# 1  NaN  5.0

# Method 4: Drop columns with missing values
print(df.dropna(axis=1))
# Output: (empty DataFrame if all columns have NaN)
```

```python
# dropna(): Removes rows or columns with missing values.
# fillna(): Replaces missing values with a specified value or method.

# Example with dropna():
df = pd.DataFrame({'A': [1, np.nan, 3], 'B': [4, 5, np.nan]})
print(df.dropna())  # Drops rows with any NaN
# Output:
#      A    B
# 0  1.0  4.0

# Example with fillna():
print(df.fillna(0))  # Fills NaN with 0
# Output:
#      A    B
# 0  1.0  4.0
# 1  0.0  5.0
# 2  3.0  0.0

# fillna() with method:
print(df.ffill())  # Forward fill
# Output:
#      A    B
# 0  1.0  4.0
# 1  1.0  5.0
# 2  3.0  5.0

# Key Differences:
# - dropna() reduces the dataset size; fillna() preserves it.
# - dropna() is for removal; fillna() is for imputation.
```

#### **19. How do you select specific columns from a DataFrame?**

```python
# You can select columns in multiple ways:

# Method 1: Using column names (single column)
df = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6], 'C': [7, 8, 9]})
print(df['A'])
# Output:
# 0    1
# 1    2
# 2    3
# Name: A, dtype: int64

# Method 2: Using column names (multiple columns)
print(df[['A', 'B']])
# Output:
#    A  B
# 0  1  4
# 1  2  5
# 2  3  6

# Method 3: Using loc for label-based selection
print(df.loc[:, ['A', 'B']])
# Output: Same as above

# Method 4: Using iloc for position-based selection
print(df.iloc[:, [0, 1]])
# Output: Same as above

# Method 5: Using filter() for pattern-based selection
print(df.filter(like='A'))
# Output:
#    A
# 0  1
# 1  2
# 2  3
```

```python
# Sorting can be done using sort_values() for one or more columns.

# Method 1: Sort by a single column (ascending)
df = pd.DataFrame({'A': [3, 1, 4, 2], 'B': [5, 6, 7, 8]})
sorted_df = df.sort_values(by='A')
print(sorted_df)
# Output:
#    A  B
# 1  1  6
# 3  2  8
# 0  3  5
# 2  4  7

# Method 2: Sort by a single column (descending)
sorted_df = df.sort_values(by='A', ascending=False)
print(sorted_df)
# Output:
#    A  B
# 2  4  7
# 0  3  5
# 3  2  8
# 1  1  6

# Method 3: Sort by multiple columns
sorted_df = df.sort_values(by=['A', 'B'], ascending=[True, False])
print(sorted_df)
# Output: (Sorts by A ascending, then B descending within same A values)

# Method 4: Sort in-place
df.sort_values(by='A', inplace=True)
print(df)
# Output: Same as Method 1
```

```python
# Scenario: Filter an e-commerce dataset to find high-value orders (amount > $1000) from specific regions.

# Example Data
data = {
    'order_id': [1, 2, 3, 4, 5],
    'region': ['East', 'West', 'East', 'South', 'West'],
    'amount': [500, 1500, 2000, 800, 3000],
    'status': ['Completed', 'Pending', 'Completed', 'Cancelled', 'Completed']
}
df = pd.DataFrame(data)

# Method 1: Boolean Indexing (Single Condition)
filtered_single = df[df['amount'] > 1000]
print("Boolean Indexing (Single Condition):\n", filtered_single)
# Output:
#    order_id region  amount     status
# 1        2   West    1500    Pending
# 2        3   East    2000  Completed
# 4        5   West    3000  Completed

# Method 2: Boolean Indexing (Multiple Conditions with &)
filtered_multi = df[(df['amount'] > 1000) & (df['region'].isin(['East', 'West']))]
print("Boolean Indexing (Multiple Conditions):\n", filtered_multi)
# Output:
#    order_id region  amount     status
# 1        2   West    1500    Pending
# 2        3   East    2000  Completed
# 4        5   West    3000  Completed

# Method 3: Using loc with Boolean Condition
filtered_loc = df.loc[df['amount'] > 1000]
print("Using loc:\n", filtered_loc)
# Output: Same as Method 1

# Method 4: Using query
filtered_query = df.query('amount > 1000')
print("Using query:\n", filtered_query)
# Output: Same as Method 1

# Method 5: Using query with Multiple Conditions
filtered_query_multi = df.query('amount > 1000 and region in ["East", "West"]')
print("Query with Multiple Conditions:\n", filtered_query_multi)
# Output: Same as Method 2

# Method 6: Using where (returns NaN for non-matching rows)
filtered_where = df.where(df['amount'] > 1000)
print("Using where:\n", filtered_where)
# Output:
#    order_id region  amount     status
# 0       NaN    NaN     NaN        NaN
# 1       2.0   West  1500.0    Pending
# 2       3.0   East  2000.0  Completed
# 3       NaN    NaN     NaN        NaN
# 4       5.0   West  3000.0  Completed

# Method 7: Using mask (opposite of where, keeps non-matching rows)
filtered_mask = df.mask(df['amount'] <= 1000)
print("Using mask:\n", filtered_mask)
# Output: Same as Method 6
```

```python
# Scenario: Merge customer profiles with their order history to analyze purchase patterns in a retail system.

# Example Data
customers = pd.DataFrame({
    'customer_id': [101, 102, 103],
    'name': ['Alice', 'Bob', 'Charlie'],
    'city': ['New York', 'Los Angeles', 'Chicago']
})
orders = pd.DataFrame({
    'customer_id': [101, 102, 104],
    'order_amount': [1000, 500, 2000],
    'order_date': ['2023-01-01', '2023-02-01', '2023-03-01']
})

# Method 1: Inner merge
merged_inner = pd.merge(customers, orders, on='customer_id', how='inner')
print("Inner merge:\n", merged_inner)
# Output:
#    customer_id     name         city  order_amount order_date
# 0         101    Alice     New York         1000 2023-01-01
# 1         102      Bob  Los Angeles          500 2023-02-01

# Method 2: Left merge
merged_left = pd.merge(customers, orders, on='customer_id', how='left')
print("Left merge:\n", merged_left)
# Output:
#    customer_id     name         city  order_amount order_date
# 0         101    Alice     New York       1000.0 2023-01-01
# 1         102      Bob  Los Angeles        500.0 2023-02-01
# 2         103  Charlie      Chicago          NaN        NaN

# Method 3: Right merge
merged_right = pd.merge(customers, orders, on='customer_id', how='right')
print("Right merge:\n", merged_right)
# Output:
#    customer_id     name         city  order_amount order_date
# 0         101    Alice     New York         1000 2023-01-01
# 1         102      Bob  Los Angeles          500 2023-02-01
# 2         104      NaN          NaN         2000 2023-03-01

# Method 4: Outer merge
merged_outer = pd.merge(customers, orders, on='customer_id', how='outer')
print("Outer merge:\n", merged_outer)
# Output:
#    customer_id     name         city  order_amount order_date
# 0         101    Alice     New York       1000.0 2023-01-01
# 1         102      Bob  Los Angeles        500.0 2023-02-01
# 2         103  Charlie      Chicago          NaN        NaN
# 3         104      NaN          NaN       2000.0 2023-03-01

# Method 5: Merge on multiple keys
orders_multi = pd.DataFrame({
    'customer_id': [101, 102, 104],
    'city': ['New York', 'Los Angeles', 'Boston'],
    'order_amount': [1000, 500, 2000]
})
merged_multi_key = pd.merge(customers, orders_multi, on=['customer_id', 'city'], how='inner')
print("Merge on multiple keys:\n", merged_multi_key)
# Output:
#    customer_id     name         city  order_amount
# 0         101    Alice     New York         1000
# 1         102      Bob  Los Angeles          500

# Method 6: Merge with index
customers_index = customers.set_index('customer_id')
orders_index = orders.set_index('customer_id')
merged_index = pd.merge(customers_index, orders_index, left_index=True, right_index=True, how='inner')
print("Merge with index:\n", merged_index)
# Output:
#              name         city  order_amount order_date
# customer_id                                            
# 101         Alice     New York         1000 2023-01-01
# 102           Bob  Los Angeles          500 2023-02-01
```

```python
# merge() combines DataFrames based on columns or indices, while join() combines based on indices.

# Example:
df1 = pd.DataFrame({'ID': [1, 2, 3], 'Name': ['Alice', 'Bob', 'Charlie']}, index=[1, 2, 3])
df2 = pd.DataFrame({'Age': [25, 30, 35]}, index=[1, 2, 4])

# Method 1: Using merge() (column-based)
merged = pd.merge(df1, df2, left_index=True, right_index=True, how='inner')
print(merged)
# Output:
#    ID     Name  Age
# 1   1    Alice   25
# 2   2      Bob   30

# Method 2: Using join() (index-based)
joined = df1.join(df2, how='inner')
print(joined)
# Output: Same as above

# Key Differences:
# - merge(): More flexible, allows merging on columns or indices, supports multiple join keys.
# - join(): Index-based, simpler for combining DataFrames with matching indices.
# - merge() supports 'on' parameter for columns; join() relies on index alignment.
```

```python
# Scenario: Transform customer ratings in a survey dataset, using apply() for complex logic and map() for simple mappings.

# Example Data
df = pd.DataFrame({
    'customer_id': [101, 102, 103],
    'rating': [3, 4, 5]
})

# Method 1: apply() with complex logic
df['rating_category'] = df['rating'].apply(lambda x: 'High' if x >= 4 else 'Low')
print("apply() with complex logic:\n", df)
# Output:
#    customer_id  rating rating_category
# 0         101       3           Low
# 1         102       4          High
# 2         103       5          High

# Method 2: map() with dictionary
rating_map = {3: 'Average', 4: 'Good', 5: 'Excellent'}
df['rating_label'] = df['rating'].map(rating_map)
print("map() with dictionary:\n", df)
# Output:
#    customer_id  rating rating_category rating_label
# 0         101       3           Low      Average
# 1         102       4          High         Good
# 2         103       5          High    Excellent

# Method 3: apply() with defined function
def categorize_rating(rating):
    return 'High' if rating >= 4 else 'Low'
df['rating_category'] = df['rating'].apply(categorize_rating)
print("apply() with defined function:\n", df)
# Output: Same as Method 1

# Method 4: map() with lambda
df['rating_label'] = df['rating'].map(lambda x: {3: 'Average', 4: 'Good', 5: 'Excellent'}[x])
print("map() with lambda:\n", df)
# Output: Same as Method 2

# Key Differences:
# - apply(): Works on Series or DataFrame, supports complex row/column operations.
# - map(): Works only on Series, best for simple element-wise mappings.
```

33. How do you select rows based on multiple conditions?

```python
# Scenario: Filter high-value customers who spent over $500 and are from the East region in a retail dataset.

# Example Data
df = pd.DataFrame({
    'customer_id': [101, 102, 103, 104],
    'region': ['East', 'West', 'East', 'South'],
    'spend': [600, 400, 800, 1000],
    'age': [25, 30, 35, 40]
})

# Method 1: Boolean indexing with & (and)
filtered_and = df[(df['spend'] > 500) & (df['region'] == 'East')]
print("Boolean indexing (and):\n", filtered_and)
# Output:
#    customer_id region  spend  age
# 0         101   East    600   25
# 2         103   East    800   35

# Method 2: Boolean indexing with | (or)
filtered_or = df[(df['spend'] > 500) | (df['age'] < 30)]
print("Boolean indexing (or):\n", filtered_or)
# Output:
#    customer_id region  spend  age
# 0         101   East    600   25
# 2         103   East    800   35
# 3         104  South   1000   40

# Method 3: Using loc with multiple conditions
filtered_loc = df.loc[(df['spend'] > 500) & (df['region'] == 'East')]
print("Using loc:\n", filtered_loc)
# Output: Same as Method 1

# Method 4: Using query
filtered_query = df.query('spend > 500 and region == "East"')
print("Using query:\n", filtered_query)
# Output: Same as Method 1

# Method 5: Using isin for list-based conditions
filtered_isin = df[df['region'].isin(['East']) & (df['spend'] > 500)]
print("Using isin:\n", filtered_isin)
# Output: Same as Method 1
```

34. What is the query() function, and how is it used?

```python
# Scenario: Query employee data to find those with high salaries and in specific departments.

# Example Data
df = pd.DataFrame({
    'emp_id': [1, 2, 3, 4],
    'department': ['HR', 'IT', 'HR', 'Sales'],
    'salary': [50000, 80000, 60000, 70000],
    'years_experience': [2, 5, 3, 4]
})

# Method 1: Basic query with single condition
filtered_single = df.query('salary > 60000')
print("Query single condition:\n", filtered_single)
# Output:
#    emp_id department  salary  years_experience
# 1      2       IT   80000                5
# 3      4    Sales   70000                4

# Method 2: Query with multiple conditions
filtered_multi = df.query('salary > 60000 and department == "IT"')
print("Query multiple conditions:\n", filtered_multi)
# Output:
#    emp_id department  salary  years_experience
# 1      2       IT   80000                5

# Method 3: Query with variable
threshold = 60000
filtered_var = df.query('salary > @threshold')
print("Query with variable:\n", filtered_var)
# Output: Same as Method 1

# Method 4: Query with in clause
filtered_in = df.query('department in ["IT", "Sales"] and salary > 60000')
print("Query with in clause:\n", filtered_in)
# Output:
#    emp_id department  salary  years_experience
# 1      2       IT   80000                5
# 3      4    Sales   70000                4

# Method 5: Query with complex expression
filtered_complex = df.query('salary / years_experience > 15000')
print("Query with complex expression:\n", filtered_complex)
# Output:
#    emp_id department  salary  years_experience
# 1      2       IT   80000                5
# 3      4    Sales   70000                4
```

35. How do you change the data type of a column?

```python
# Scenario: Convert transaction dates to datetime and amounts to float in a financial dataset.

# Example Data
df = pd.DataFrame({
    'transaction_id': [1, 2, 3],
    'date': ['2023-01-01', '2023-02-01', '2023-03-01'],
    'amount': ['1000.50', '2000.75', '1500.25']
})

# Method 1: Using astype to convert to float
df['amount'] = df['amount'].astype(float)
print("Using astype to float:\n", df.dtypes)
# Output:
# transaction_id      int64
# date              object
# amount           float64
# dtype: object

# Method 2: Using to_numeric for amount
df['amount'] = pd.to_numeric(df['amount'], errors='coerce')
print("Using to_numeric:\n", df.dtypes)
# Output: Same as Method 1

# Method 3: Convert date to datetime
df['date'] = pd.to_datetime(df['date'])
print("Convert date to datetime:\n", df.dtypes)
print(df)
# Output:
# transaction_id             int64
# date              datetime64[ns]
# amount                   float64
# dtype: object

# Method 4: Using astype for categorical
df['transaction_id'] = df['transaction_id'].astype(str)
print("Convert to string (categorical):\n", df.dtypes)
# Output:
# transaction_id            object
# date              datetime64[ns]
# amount                   float64
# dtype: object

# Method 5: Using convert_dtypes for optimal types
df = df.convert_dtypes()
print("Using convert_dtypes:\n", df.dtypes)
# Output:
# transaction_id    string[python]
# date              datetime64[ns]
# amount                   Float64
# dtype: object
```

38. How do you reset the index of a DataFrame?

40. What is the difference between drop() and del?

```python
# Scenario: Clean a marketing campaign dataset by removing irrelevant columns or rows.

# Example Data
df = pd.DataFrame({
    'campaign_id': [1, 2, 3],
    'clicks': [100, 200, 300],
    'impressions': [1000, 2000, 3000],
    'conversions': [10, 20, 30]
})

# Method 1: drop columns
df_drop_col = df.drop(columns='impressions')
print("Drop column:\n", df_drop_col)
# Output:
#    campaign_id  clicks  conversions
# 0           1     100          10
# 1           2     200          20
# 2           3     300          30

# Method 2: drop multiple columns
df_drop_multi = df.drop(columns=['impressions', 'conversions'])
print("Drop multiple columns:\n", df_drop_multi)
# Output:
#    campaign_id  clicks
# 0           1     100
# 1           2     200
# 2           3     300

# Method 3: drop rows
df_drop_row = df.drop(index=0)
print("Drop row:\n", df_drop_row)
# Output:
#    campaign_id  clicks  impressions  conversions
# 1           2     200         2000          20
# 2           3     300         3000          30

# Method 4: drop in-place
df.drop(columns='impressions', inplace=True)
print("Drop in-place:\n", df)
# Output:
#    campaign_id  clicks  conversions
# 0           1     100          10
# 1           2     200          20
# 2           3     300          30

# Method 5: Using del
del df['conversions']
print("Using del:\n", df)
# Output:
#    campaign_id  clicks
# 0           1     100
# 1           2     200
# 2           3     300

# Key Differences:
# - drop(): Removes rows or columns, supports non-inplace operations, flexible for multiple labels.
# - del: Removes columns only, always in-place, more memory-efficient for single columns.
```
