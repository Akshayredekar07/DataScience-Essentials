```python
def linear_regression(X, y, lr=0.001, max_iteration=1000):
    # y=mx+b
    m, b = 0, 0
    n = len(X)

    for _ in range(max_iteration):
        y_pred = m*X + b
        loss = y - y_pred

        d_m = (-2/n)*sum(X*loss)
        d_b = (-2/n)*sum(loss)


        m =  m - lr*d_m
        b =  b - lr*d_b

    return m, b
```

```python
# Sample dataset
X = [1400, 1600, 1700, 1875, 1100, 1550, 2350, 2450, 1425, 1700]  # Size
y = [245, 312, 279, 308, 199, 219, 405, 324, 319, 255]           # Price

# Initialize parameters
w = 0.0  # Slope
b = 0.0  # Intercept
alpha = 0.001  # Learning rate
iterations = 1000
n = len(X)

# Gradient Descent
for _ in range(iterations):
    # Compute predictions
    y_pred = [w * x + b for x in X]
    
    # Compute gradients
    dw = (1/n) * sum([x * (y_pred[i] - y[i]) for i, x in enumerate(X)])
    db = (1/n) * sum([y_pred[i] - y[i] for i in range(n)])
    
    # Update parameters
    w -= alpha * dw
    b -= alpha * db

# Print results
print(f"Slope (w): {w:.4f}, Intercept (b): {b:.4f}")

# Make a prediction (e.g., for size = 1500 sqft)
size = 1500
price = w * size + b
print(f"Predicted price for {size} sqft: ${price:.2f}k")

# Compute Mean Squared Error
y_pred = [w * x + b for x in X]
mse = sum([(y_pred[i] - y[i])**2 for i in range(n)]) / n
print(f"Mean Squared Error: {mse:.2f}")
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import numpy as np

# Sample dataset
X = np.array([1400, 1600, 1700, 1875, 1100, 1550, 2350, 2450, 1425, 1700]).reshape(-1, 1)
y = np.array([245, 312, 279, 308, 199, 219, 405, 324, 319, 255])

# Initialize and fit the model
model = LinearRegression()
model.fit(X, y)

# Get parameters
w = model.coef_[0]
b = model.intercept_
print(f"Slope (w): {w:.4f}, Intercept (b): {b:.4f}")

# Make a prediction (e.g., for size = 1500 sqft)
size = np.array([[1500]])
price = model.predict(size)[0]
print(f"Predicted price for 1500 sqft: ${price:.2f}k")

# Compute Mean Squared Error
y_pred = model.predict(X)
mse = mean_squared_error(y, y_pred)
print(f"Mean Squared Error: {mse:.2f}")
```

```python
# Linear Regression from scratch using Gradient Descent

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# Initialize parameters
w = 0
b = 0
alpha = 0.01
epochs = 1000
n = len(x)

# Training
for epoch in range(epochs):
    y_pred = [w * xi + b for xi in x]
    dw = -(2/n) * sum((yi - ypi) * xi for xi, yi, ypi in zip(x, y, y_pred))
    db = -(2/n) * sum(yi - ypi for yi, ypi in zip(y, y_pred))

    w -= alpha * dw
    b -= alpha * db

# Final model
print(f"Learned model: y = {w:.2f}x + {b:.2f}")
```

```python
x_test = np.array([6, 7, 8])
y_test_pred = w * x_test + b
print("Predicted:", y_test_pred)
```

```python
import numpy as np

# Sample data
x = np.array([1, 2, 3, 4, 5], dtype=np.float64)
y = np.array([2, 4, 6, 8, 10], dtype=np.float64)

# Initialize parameters
w = 0.0  # slope
b = 0.0  # intercept
lr = 0.01
epochs = 1000
n = len(x)

# Gradient Descent Loop
for epoch in range(epochs):
    y_pred = w * x + b
    error = y_pred - y
    
    # Gradients
    dw = (2/n) * np.dot(error, x)
    db = (2/n) * np.sum(error)
    
    # Update parameters
    w -= lr * dw
    b -= lr * db

# Final model
print(f"Learned equation: y = {w:.2f}x + {b:.2f}")
```

```python
x_test = np.array([6, 7, 8])
y_test_pred = w * x_test + b
print("Predicted:", y_test_pred)
```
