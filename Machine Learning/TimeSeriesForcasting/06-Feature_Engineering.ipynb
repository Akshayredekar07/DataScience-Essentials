
**Feature Engineering for Time-Series Data**

Time-series data is sequential: past values influence future outcomes. Standard ML models do not understand time, so **feature engineering** is necessary.

Core techniques:

1. **Lag Features** – include past values.
2. **Rolling Window Features** – capture trends and volatility.
3. **Sine-Cosine Encoding** – encode cyclical time features properly.



**1. Lag Features**

**Purpose:** Give the model memory of past values.

**Example:** Predict electricity demand: yesterday’s or last week’s demand influences today.

**Python Implementation:**

```python
import pandas as pd

# Assuming df has 'value' column
df["lag_1"] = df["value"].shift(1)    # previous day
df["lag_7"] = df["value"].shift(7)    # one week ago
df["lag_30"] = df["value"].shift(30)  # one month ago

df.dropna(inplace=True)  # remove NaNs created by shifting
```

**Impact:**

* Models learn trends and seasonality.
* Captures weekly/monthly patterns.

**Analogy:** Stock price forecasting using historical prices for momentum or mean reversion patterns.



**2. Rolling Window Features**

**Purpose:** Capture trends and volatility over time.

**Example:** 7-day average temperature or 30-day electricity demand volatility.

**Python Implementation:**

```python
# Moving averages (trend)
df["rolling_mean_7"] = df["value"].rolling(window=7).mean()
df["rolling_mean_30"] = df["value"].rolling(window=30).mean()

# Rolling standard deviation (volatility)
df["rolling_std_7"] = df["value"].rolling(window=7).std()
df["rolling_std_30"] = df["value"].rolling(window=30).std()

df.dropna(inplace=True)
```

**Impact:**

* Moving averages show direction of trend.
* Rolling std indicates stability/volatility of recent values.

**Analogy:** Risk management in finance uses rolling volatility to adjust portfolios.


**3. Sine-Cosine Encoding for Cyclical Features**

**Purpose:** Encode cyclical features (hours, days, months) to preserve continuity.

**Problem:** Integer encoding (e.g., Monday=0, Sunday=6) is misleading because 6 → 0 seems far apart.

**Python Implementation:**

```python
import numpy as np

# Example: day of the week (0=Monday, 6=Sunday)
df["sin_day_of_week"] = np.sin(2 * np.pi * df["day_of_week"] / 7)
df["cos_day_of_week"] = np.cos(2 * np.pi * df["day_of_week"] / 7)
```

**Impact:**

* Preserves cyclic nature.
* Improves tree-based model learning.

**Analogy:** Signal processing uses sine/cosine to represent periodic patterns efficiently.



**Summary**

| Feature Type         | Purpose                                | Key Benefit                      |

| Lag Features         | Include past values                    | Captures history and seasonality |
| Rolling Window       | Compute mean/std over a window         | Captures trends and volatility   |
| Sine-Cosine Encoding | Encode cyclical features (hours, days) | Maintains cyclic continuity      |



**Python Notes:**

* Always `dropna()` after shifts or rolling calculations.

**1. Lag Features – Giving the Model Memory**

**Purpose:** Allow the model to see past values of the target variable.

**Formula / Implementation:**

$$
\text{lag}_k = Y_{t-k}
$$

**Python Example:**

```python
df["lag_1"] = df["value"].shift(1)   # 1-step lag
df["lag_7"] = df["value"].shift(7)   # 7-day lag
df["lag_30"] = df["value"].shift(30) # 30-day lag
df.dropna(inplace=True)
```

**Effect:** Helps models capture short-term trends, weekly patterns, or monthly seasonality.

**Analogy:** Like a trader using historical stock prices to predict tomorrow’s price.


**2. Rolling Window Features – Trends and Volatility**

**Purpose:** Summarize past values to capture trends (mean) and volatility (standard deviation).

**Formulas:**

$$
\text{Rolling Mean}_k = \frac{1}{k}\sum_{i=0}^{k-1} Y_{t-i}
$$

$$
\text{Rolling Std}_k = \sqrt{\frac{1}{k}\sum_{i=0}^{k-1} (Y_{t-i} - \text{Rolling Mean}_k)^2}
$$

**Python Example:**

```python
# Trend
df["rolling_mean_7"] = df["value"].rolling(window=7).mean()
df["rolling_mean_30"] = df["value"].rolling(window=30).mean()

# Volatility
df["rolling_std_7"] = df["value"].rolling(window=7).std()
df["rolling_std_30"] = df["value"].rolling(window=30).std()
df.dropna(inplace=True)
```

**Effect:** Helps model recognize increasing/decreasing trends and fluctuating values.

**Analogy:** Like a risk manager monitoring a stock’s rolling volatility to adjust exposure.


**3. Sine-Cosine Encoding – Handling Cyclical Features**

**Purpose:** Represent cyclic time features (hour, day, month) so models understand adjacency.

**Formulas:**

$$
\text{sin}_t = \sin\left(2 \pi \frac{t}{T}\right), \quad
\text{cos}_t = \cos\left(2 \pi \frac{t}{T}\right)
$$

where $T$ = cycle length (e.g., 7 for day-of-week).

**Python Example:**

```python
import numpy as np
df["sin_day_of_week"] = np.sin(2 * np.pi * df["day_of_week"] / 7)
df["cos_day_of_week"] = np.cos(2 * np.pi * df["day_of_week"] / 7)
```

**Effect:** Ensures cyclic continuity (Sunday → Monday) is preserved.

**Analogy:** Like breaking signals into sine/cosine waves in signal processing to capture repeating patterns.

| Sine-Cosine Encode | Cyclical feature representation | `df['sin_day'] = np.sin(2*np.pi*df['day']/7)`            |

**Takeaway:** These transformations make sequential patterns explicit, allowing tree-based or even deep learning models to learn temporal relationships effectively.
