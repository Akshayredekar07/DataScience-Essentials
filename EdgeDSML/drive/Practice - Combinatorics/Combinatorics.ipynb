### **Q1. Given three T/F questions. Count the total ways to answer the questions.**  

#### **Solution:**  

For each question, there are **2 possible answers**: **True (T) or False (F)**.  

Since there are **3 questions**, we can find the total number of ways to answer them by multiplying the possibilities for each question:  

$$
2 \text{ (options for Q1)} \times 2 \text{ (options for Q2)} \times 2 \text{ (options for Q3)} = 2^3 = 8
$$

**Therefore, there are 8 total ways to answer the three True/False questions.**

| Question 1 | Question 2 | Question 3 |

| T          | T          | F          |  
| T          | F          | T          |  
| T          | F          | F          |  
| F          | T          | T          |  
| F          | T          | F          |  
| F          | F          | T          |  
| F          | F          | F          |  



### **Q2. 10 Girls and 7 Boys in a Hall. Count the number of ways to pick 1 boy and 1 girl.**  

#### **Solution:**  

- There are **10 choices** for selecting **1 girl** out of **10 girls**.  
- There are **7 choices** for selecting **1 boy** out of **7 boys**.  
- To find the **total number of ways** to pick **1 girl and 1 boy**, we multiply the number of choices:  

$$
10 \times 7 = 70
$$

**Therefore, there are 70 ways to pick 1 girl and 1 boy from the group.**  


```python
girls = [f"G{i}" for i in range(1, 11)]  # G1 to G10
boys = [f"B{i}" for i in range(1, 8)]   # B1 to B7

pairings = [(girl, boy) for girl in girls for boy in boys]

# pairings = []  # Empty list to store pairs
# # Using nested for loops to generate all pairs
# for girl in girls:
#     for boy in boys:
#         pairings.append((girl, boy))


print("Some possible pairings:")
for pair in pairings[:10]: 
    print(pair)

print(f"\nTotal number of ways to pick 1 boy and 1 girl: {len(pairings)}")
```
