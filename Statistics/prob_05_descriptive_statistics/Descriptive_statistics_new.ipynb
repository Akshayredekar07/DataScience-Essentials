```python
# Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.distributions.empirical_distribution import ECDF

# Load Sehwag's dataset
sehwag = pd.read_csv("sehwag.csv")

# Display the first few rows to understand the data
print(sehwag.head())
```

```python
# Check the number of rows and columns
print(sehwag.shape)
```

```python
# Get summary statistics for the Runs column
print(sehwag["Runs"].describe())
```

```python
# Calculate mean
mean_runs = sehwag["Runs"].mean()
print(f"Mean Runs: {mean_runs:.2f}")
```

```python
# Calculate median
median_runs = sehwag["Runs"].median()
print(f"Median Runs: {median_runs:.2f}")
```

```python
# Calculate mode
mode_runs = sehwag["Runs"].mode()[0]
print(f"Mode Runs: {mode_runs}")
```

```python
# Calculate range
min_runs = sehwag["Runs"].min()
max_runs = sehwag["Runs"].max()
range_runs = max_runs - min_runs
print(f"Min Runs: {min_runs}, Max Runs: {max_runs}, Range: {range_runs}")
```

```python
# Calculate variance
variance_runs = sehwag["Runs"].var()
print(f"Variance: {variance_runs:.2f}")
```

```python
# Calculate standard deviation
std_runs = sehwag["Runs"].std()
print(f"Standard Deviation: {std_runs:.2f}")
```

```python
# Calculate Q1, Q3, and IQR
q1 = sehwag["Runs"].quantile(0.25)  # 25th percentile
q3 = sehwag["Runs"].quantile(0.75)  # 75th percentile
iqr = q3 - q1
print(f"Q1 (25th Percentile): {q1}")
print(f"Q3 (75th Percentile): {q3}")
print(f"IQR: {iqr}")
```

```python
# Calculate outlier bounds
lower_bound = max(q1 - 1.5 * iqr, 0)  # Runs can't be negative
upper_bound = q3 + 1.5 * iqr
print(f"Lower Bound: {lower_bound}")
print(f"Upper Bound: {upper_bound}")

# Identify outliers
outliers = sehwag["Runs"][(sehwag["Runs"] < lower_bound) | (sehwag["Runs"] > upper_bound)]
print(f"Outliers: {outliers.values}")
print(f"Number of Outliers: {len(outliers)}")
```

```python
# Plot histogram
plt.figure(figsize=(8, 4))
sns.histplot(data=sehwag, x="Runs", bins=20, kde=True, color='green', alpha=0.9)
plt.title("Histogram of Sehwag's Runs", fontsize=10)
plt.xlabel("Runs", fontsize=8)
plt.ylabel("Frequency", fontsize=8)
plt.xticks(fontsize=8)
plt.yticks(fontsize=8)
plt.show()
```

```python
# Plot box plot of Sehwag's run distribution
plt.figure(figsize=(10, 6))
sns.boxplot(x=sehwag["Runs"], color='skyblue')
plt.title("Box Plot of Sehwag's Runs Distribution", fontsize=12, pad=15)
plt.xlabel("Runs Scored", fontsize=10)
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# Calculate and plot Empirical Cumulative Distribution Function (ECDF)
ecdf = ECDF(sehwag["Runs"])
plt.figure(figsize=(12, 8))
plt.plot(ecdf.x, ecdf.y, label="ECDF", color='navy', linewidth=2)
plt.title("Empirical Cumulative Distribution Function of Sehwag's Runs", 
          fontsize=14, pad=15)
plt.xlabel("Runs Scored", fontsize=12)
plt.ylabel("Cumulative Probability", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(fontsize=10)
plt.tight_layout()
plt.show()
```
