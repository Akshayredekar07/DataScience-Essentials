```python
with open("smartprix.html", "r", encoding="utf-8") as f:
    html = f.read()
```

```python
from bs4 import BeautifulSoup

soup = BeautifulSoup(html, 'lxml')
soup.prettify()
```

```python
containers = soup.find_all('div', class_="sm-product has-tag has-features has-actions")
```

```python
len(containers)
```

```python
for container in containers:
    h2_elements = container.find_all('h2')
    for h2 in h2_elements:
        print(h2.text)
```

```python
for item in containers:
    print(item.find_all('h2'))
```

```python
for item in containers:
    phones = item.find_all('h2')
    for phone in phones:
        print(phone.text)
```
