```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.common.keys import Keys
import time
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options

# Set up Chrome options
options = Options()
options.add_argument("--start-maximized")
options.add_argument("--disable-notifications")

CHROMEDRIVER_PATH = "C:/Users/Admin/Downloads/APPS/chromedriver.exe"  # Update this path
DEPARTURE_CITY = "Mumbai"
ARRIVAL_CITY = "Delhi"
SOURCE_AIRPORT = "BOM"
DESTINATION_AIRPORT = "DEL"
DEPARTURE_DATE = "02/08/2025" 


# Initialize the WebDriver
service = Service("C:/Users/Admin/Downloads/APPS/chromedriver.exe")
driver = webdriver.Chrome(service=service, options=options)

# Open Yatra website
driver.get("https://www.makemytrip.com/")
time.sleep(2)

# Initialize WebDriverWait
wait = WebDriverWait(driver, 4)

try:
    close_btn1 = wait.until(EC.element_to_be_clickable((By.XPATH, '//*[@id="SW"]/div[1]/div[2]/div[2]/div/section/span')))
    close_btn1.click()
except Exception as e:
    print("Login pop-up not found or already closed:", e)


driver.find_element(By.XPATH, "//span[text()='Flights']").click()
time.sleep(2)

# Select departure city (Mumbai)
from_input = WebDriverWait(driver, 10).until(EC.element_to_be_clickable((By.ID, "fromCity")))
from_input.click()
from_input.send_keys(DEPARTURE_CITY)
WebDriverWait(driver, 10).until(EC.element_to_be_clickable((By.XPATH, f"//p[contains(text(),'{DEPARTURE_CITY}')]"))).click()
time.sleep(2)

# Select arrival city (Delhi)
to_input = WebDriverWait(driver, 10).until(EC.element_to_be_clickable((By.ID, "toCity")))
to_input.click()
to_input.send_keys(ARRIVAL_CITY)
WebDriverWait(driver, 10).until(EC.element_to_be_clickable((By.XPATH, f"//p[contains(text(),'{ARRIVAL_CITY}')]"))).click()
time.sleep(2)  # Pause to stabilize page


#***************************************************************
time.sleep(4)
# Suppose you want to click on 2nd August 2025
target_date = "Sat Aug 02 2025"

# Click the calendar date using aria-label
driver.find_element(By.XPATH, f"//div[@aria-label='{target_date}']").click()


search_btn = WebDriverWait(driver, 2).until(
    EC.element_to_be_clickable((By.XPATH, "//a[contains(@class, 'widgetSearchBtn')]"))
)
driver.execute_script("arguments[0].click();", search_btn)
```
