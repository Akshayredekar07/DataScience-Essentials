```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import TimeoutException, NoSuchElementException
import time
from datetime import datetime, timedelta
import pandas as pd

# === Setup ===
CHROMEDRIVER_PATH = "C:/Users/Admin/Downloads/APPS/chromedriver.exe"
DEPARTURE_CITY = "Mumbai"
ARRIVAL_CITY = "Delhi"
SOURCE_AIRPORT = "BOM"
DEST_AIRPORT = "DEL"
LAYOVER = "non-stop"
DEPARTURE_DATE = (datetime.today() + timedelta(days=7)).strftime('%d/%m/%Y')

options = Options()
options.add_argument("--start-maximized")
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_argument("--disable-notifications")

service = Service(CHROMEDRIVER_PATH)
driver = webdriver.Chrome(service=service, options=options)

# === Open MakeMyTrip and handle popups ===
driver.get("https://www.makemytrip.com/")
time.sleep(5)

# Close login popup
try:
    driver.find_element(By.XPATH, '//*[@id="SW"]/div[1]/div[2]/div[2]/div/section/span').click()
# === Input Source City ===
wait = WebDriverWait(driver, 10)

# Click source city field
wait.until(EC.element_to_be_clickable((By.ID, "fromCity"))).click()
time.sleep(2)

# Clear and enter source city
from_input = wait.until(EC.presence_of_element_located((By.XPATH, "//input[@placeholder='From']")))
from_input.clear()
from_input.send_keys(DEPARTURE_CITY)
time.sleep(2)

# Wait for suggestions to appear and click the city
try:
    # Try multiple possible selectors for the city option
    city_option = wait.until(EC.element_to_be_clickable((By.XPATH, f"//p[contains(text(),'{DEPARTURE_CITY}')]")))
    city_option.click()
except TimeoutException:
    try:
        city_option = wait.until(EC.element_to_be_clickable((By.XPATH, f"//span[contains(text(),'{DEPARTURE_CITY}')]")))
        city_option.click()
    except TimeoutException:
        # Try clicking the first suggestion
        suggestions = driver.find_elements(By.XPATH, "//ul[@class='react-autosuggest__suggestions-list']//li")
        if suggestions:
            suggestions[0].click()

# === Input Destination City ===
time.sleep(2)
wait.until(EC.element_to_be_clickable((By.ID, "toCity"))).click()
time.sleep(2)

# Clear and enter destination city
to_input = wait.until(EC.presence_of_element_located((By.XPATH, "//input[@placeholder='To']")))
to_input.clear()
to_input.send_keys(ARRIVAL_CITY)
time.sleep(2)

# Wait for suggestions to appear and click the city
try:
    # Try multiple possible selectors for the city option
    city_option = wait.until(EC.element_to_be_clickable((By.XPATH, f"//p[contains(text(),'{ARRIVAL_CITY}')]")))
    city_option.click()
except TimeoutException:
    try:
        city_option = wait.until(EC.element_to_be_clickable((By.XPATH, f"//span[contains(text(),'{ARRIVAL_CITY}')]")))
        city_option.click()
    except TimeoutException:
        # Try clicking the first suggestion
        suggestions = driver.find_elements(By.XPATH, "//ul[@class='react-autosuggest__suggestions-list']//li")
        if suggestions:
            suggestions[0].click()
driver.find_element(By.XPATH, "//input[@placeholder='To']").send_keys(ARRIVAL_CITY)
time.sleep(1)
driver.find_element(By.XPATH, f"//p[text()='{ARRIVAL_CITY}']").click()

# === Set departure date ===
driver.find_element(By.XPATH, "//label[@for='departure']").click()
time.sleep(1)

# Click correct date from calendar
target_date = datetime.today() + timedelta(days=7)
formatted_label = target_date.strftime('%a %b %d %Y')  # e.g., Thu Aug 07 2025
try:
    driver.find_element(By.XPATH, f"//div[@aria-label='{formatted_label}']").click()
except:
    print("Could not find date. Please manually inspect and correct selector.")
    driver.quit()
    exit()

# === Search Flights ===
driver.find_element(By.XPATH, "//a[text()='Search']").click()
time.sleep(15)  # wait for results to load

# === Extract Flight Data ===
flights = driver.find_elements(By.XPATH, "//div[@class='listingCard']")
flight_data = []

for flight in flights:
    try:
        airline = flight.find_element(By.XPATH, ".//span[@class='airlineInfo-sctn']").text.strip()
        time_departure = flight.find_element(By.XPATH, ".//div[@class='timeInfoLeft']/p").text.strip()
        price_text = flight.find_element(By.XPATH, ".//div[@class='blackText fontSize18 blackFont white-space-no-wrap']").text.replace(",", "")
        total_price = int(price_text)

        # If flight number available
        try:
            flight_number = flight.find_element(By.XPATH, ".//span[contains(@class,'flightNo')]").text.strip()
        except:
            flight_number = airline  # fallback

        flight_data.append({
            "Flight_Number": flight_number,
            "Source_City": DEPARTURE_CITY,
            "Destination_City": ARRIVAL_CITY,
            "Source_Airport": SOURCE_AIRPORT,
            "Destination_Airport": DEST_AIRPORT,
            "Date": DEPARTURE_DATE,
            "Base_Fare": "",  # Optional: parsing needed from fare breakdown
            "Tax": "",        # Optional: parsing needed from fare breakdown
            "Total_Fare": total_price,
            "Layover": LAYOVER
        })
    except Exception as e:
        print("Skipping a flight due to parsing error:", e)
        continue

# === Save to Excel ===
df = pd.DataFrame(flight_data)
df.to_excel("output_flights.xlsx", index=False)

print("Saved:", df.shape[0], "flights.")
driver.quit()
```
